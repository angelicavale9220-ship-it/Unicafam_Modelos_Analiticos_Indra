***FUNCIONES***

```python
# CALCULAR

def calcular_factura(subtotal, tasa_iva=0.19, descuento=0):

    """Devuelve el total a pagar de una factura."""

    neto = subtotal * (1 - descuento)

    iva = neto * tasa_iva

    return round(neto + iva)
```

```python
# 1 IVA del 19% y sin descuento

print(calcular_factura(100000))
```

```python

# 2 IVA del 19% y descuento del 10%

print(calcular_factura(100000, descuento=0.1))
```

```python
# 3 IVA del 5% y sin descuento

print(calcular_factura(100000, tasa_iva=0.05))
```

***CLASIFICACION DE NOTAS***

```python
# NOTAS

def clasificar_nota(nota):

    if nota < 0 or nota > 5:
        return "Nota inválida"

    elif nota < 3.0:
        return "Reprueba"

    elif nota < 4.5:
        return "Aprueba"

    else:
        return "Excelente"
```

```python
# Lista de notas

notas = [2.5, 3.8, 4.7, 5.2, 3.0]
```

```python
# Inicializar el contador

aprobados = 0
```

```python
# Aplicar la función a cada nota

for nota in notas:
    resultado = clasificar_nota(nota)
    print(nota, "→", resultado)

    if resultado == "Aprueba" or resultado == "Excelente":
        aprobados += 1
```

```python
print("Total de estudiantes aprobados:", aprobados)
```

***NOMINA CON PANDAS ***

```python
import pandas as pd

import numpy as np
```

```python
# 1 Tabla inicial de empleados

df = pd.DataFrame({

    "empleado": ["Ana", "Luis", "Marta", "Juan", "Sofía"],
    "area": ["Ventas", "TI", "Ventas", "TI", "RR. HH."],
    "salario": [2800000, 4500000, 3100000, 5200000, 3000000]})
```

```python
# 2 Agregar empleados nuevos UNA SOLA VEZ

nuevos = pd.DataFrame({

    "empleado": ["Pedro", "Lina", "Camilo"],
    "area": ["Ventas", "TI", "Ventas"],
    "salario": [2900000, 4800000, 3300000]})
```

```python
# 3 Unir las tablas

df = pd.concat([df, nuevos], ignore_index=True)
```

```python
# 4 Calcular bonos

df["bono"] = np.where(

    df["area"] == "Ventas",
    df["salario"] * 0.10,
    df["salario"] * 0.05)
```

```python
# 5 Calcular costo

df["costo"] = df["salario"] + df["bono"]
```

```python
# 6 Mostrar costo por área

print(df.groupby("area")["costo"].sum())
```

```python
# 7 Encontrar mayor salario

print("Mayor salario:",
      df.loc[df["salario"].idxmax(), "empleado"])
```

```python
# 8 Mostrar tabla final

print("\nTABLA FINAL")

print(df)
```
