```python
def ventas_mes(mes):
    datos = {
        "ene": [100, 140, 130],
        "feb": [120, 150, 130],
        "mar": [134, 145, 131]
    }

    return pd.DataFrame({
        "mes": mes,
        "sucursal": ["Norte", "Sur", "Centro"],
        "Ventas": np.array(datos[mes]) * 1e6
    })
```

```python
ene = ventas_mes("ene")
feb = ventas_mes("feb")
mar = ventas_mes("mar")
```

```python
Consolidado = pd.concat(
    [ene, feb, mar],
    ignore_index=True
)
```

```python
resumen = Consolidado.groupby("sucursal").agg(
    total=("Ventas", "sum"),
    promedio=("Ventas", "mean"),
    meses=("mes", "count")
)

resumen = (resumen / [1e6, 1e6, 1]).round(1)

print(resumen.sort_values("total", ascending=False))
```

***INTEGRAR CON SQL***

```python
import sqlite3
import pandas as pd
```

```python

# 1 Crear una base de datos en memoria

con = sqlite3.connect(":memory:")
```

```python
# 2 Guardar los DataFrames como tablas SQL

clientes.to_sql("clientes", con, index=False)
```

```python
pedidos.rename(
    columns={"pago.valor": "valor"}
)[["pedido", "cliente", "valor"]].to_sql(
    "pedidos", con, index=False
)
```

```python
# 3 Crear la consulta SQL

query = """
SELECT
    c.ciudad,
    COUNT(p.pedido) AS pedidos,
    SUM(p.valor) AS ventas
FROM clientes c
LEFT JOIN pedidos p
    ON c.id_cliente = p.cliente
GROUP BY c.ciudad
ORDER BY ventas DESC
"""
```

```python
# 4 Ejecutar la consulta y mostrar resultados

print(pd.read_sql(query, con))
```
